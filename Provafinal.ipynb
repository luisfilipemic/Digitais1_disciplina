#### Analise de saídas: com base em soma de produtos;

x3 será sempre baixa;


x2 será alta nas condições de entrada 0000;

x1 será alta nas condições de entrada 0001 e 0011;

x0 será alta nas condições de entrada 0001 e 0111;

- Portanto:

x3 = sempre zero;

x2 = 0000;

x1 = 0001 + 0011;

x0 = 0001 + 0111.
